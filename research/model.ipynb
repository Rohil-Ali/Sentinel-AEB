# Model Testing

```python
from ultralytics import YOLO

model = YOLO('yolov8s.pt', )  # loads pre trained model on coco
```

## Baseline testing 

```python
metrics = model.val(data='dataset/data.yaml')

print(f"mAP50-95: {metrics.box.map}") 
print(f"mAP50: {metrics.box.map50}")
```

## Retrain model

```python
#Train on BDD100K
model.train(
    data='dataset/data.yaml', 
    epochs=10, 
    imgsz=640, 
    batch=16,
    freeze=10,
    name='yolov8s_bdd_finetune'
)
```
